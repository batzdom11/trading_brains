```python
from google.cloud import bigquery
import pandas as pd
# Initialize client (uses your gcloud credentials)
client = bigquery.Client(project='trading-brains')

# Your query
query = """
SELECT 
  p.timestamp,
  p.pred_15m, a.actual_15m_price, (a.actual_15m_price - p.pred_15m) as error_15m,
  p.pred_30m, a.actual_30m_price, (a.actual_30m_price - p.pred_30m) as error_30m,
  p.pred_45m, a.actual_45m_price, (a.actual_45m_price - p.pred_45m) as error_45m,
  p.pred_60m, a.actual_60m_price, (a.actual_60m_price - p.pred_60m) as error_60m
FROM `trading-brains.tft_predictions.tft_predictions_logs` p
JOIN `trading-brains.tft_predictions.tft_actuals` a
  ON p.timestamp = a.prediction_timestamp
ORDER BY p.timestamp DESC
"""

# Run query and get DataFrame
df = client.query(query).to_dataframe()
df
```
